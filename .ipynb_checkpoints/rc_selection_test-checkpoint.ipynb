```python
# Import everything from rc_selection script 
import scripts.rc_selection as rc
```

```python
# Defined Parameters 
field_name = 'MISHAPS_F1_DITH0'  # Field identifier
chip_id = 'N9'  # Chip identifier
magnitude_band = 'r'  # Band used for magnitude measurement
color_band_1 = 'r'  # First band for color (e.g. r-z)
color_band_2 = 'z'  # Second band for color
field_chip_label = f"{field_name}_{chip_id}"  # Combined label for file references

# Define file path
file_path = "Data_Files/" + field_chip_label + ".cmd.calibrate_ps"
print(f"Reading catalog from: {file_path}")
```

```python
# Load in file and apply photometric filtering 
data = rc.filtered_magnitude(file_path, err_cut=0.2)

print(f"Stars after quality cuts: {len(data)}")
```

```python
rc_model = rc.initialize_red_clump(data)

clump_r = rc_model["clump-r"]
clump_z = rc_model["clump-z"]

print(f"Predicted RC location: r = {clump_r:.2f}, r−z = {clump_r - clump_z:.2f}")
```
